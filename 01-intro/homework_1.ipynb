## P6. Valor mediano de caballos de fuerza (horsepower)

```python
hp_col = df_Car['horsepower']
mediana_antes = hp_col.median()
mediana_antes
```

```python
moda_hp = hp_col.mode()[0]
moda_hp
```

```python
hp_rellenado = hp_col.fillna(moda_hp)
mediana_despues = hp_rellenado.median()
mediana_despues
```

```python
print(f"Antes: {mediana_antes:.1f}, Despues: {mediana_despues:.1f}")
if mediana_antes == mediana_despues:
    print("No cambio")
elif mediana_despues > mediana_antes:
    print("Yes, it increased")
else:
    print("Yes, it decreased")
```

## P7. Suma de los pesos

```python
autos_asia = df_Car[df_Car['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)
X = autos_asia.to_numpy()
X
```

```python
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
w
```
